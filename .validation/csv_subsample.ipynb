```python
import pandas as pd
import random
import os.path
```

```python
seed = 42
n    = 30  # Number of rows to subsample from each file

# Set a fixeàd random seed for reproducibility
random.seed(seed)  # You can use any number (e.g., 123, 2026)

# File paths
p     = "/home/estopinj/POSTDOC/OBSGESSION/WP3/DAM Framework/Validation"
file1 = os.path.join(p, "Dimensions-Publication-2026-08-26_08-43-22.csv") # Detection
file2 = os.path.join(p, "Dimensions-Publication-2026-08-26_08-34-46.csv") # Attribution

# Output filenames
output1 = os.path.join(p, f"detection_subsample_seed{seed}_n{n}.csv")
output2 = os.path.join(p, f"attribution_subsample_seed{seed}_n{n}.csv")

# Subsample 30 rows from each file
def subsample_csv(input_file, output_file, n=n):
    df = pd.read_csv(input_file, skiprows=1)
    print(f"Read {len(df)} rows from {input_file}")
    print(df.head())  # Print the first few rows for verification
    if len(df) < n:
        raise ValueError(f"File {input_file} has fewer than {n} rows.")
    sample = df.sample(n=n, random_state=seed)  # random_state ensures reproducibility
    sample.to_csv(output_file, index=False)
    print(f"Saved {n} rows to {output_file}")

# Process both files
subsample_csv(file1, output1)
subsample_csv(file2, output2)
```
